```python
dna_sequence = "ATGCGATCGATCGATCGATCGATCGACTGACTAGCTAGCTAGC"

gc_count = dna_sequence.count("G") + dna_sequence.count("C")
gc_percentage = (gc_count / len(dna_sequence)) * 100

print(f"Total DNA Length: {len(dna_sequence)} bases")
print(f"Calculated GC Content: {gc_percentage:.2f}%")
```
